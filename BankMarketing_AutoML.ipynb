# Bank Marketing - AutoML

## Introduction

This notebook demonstrates how to apply AutoML using PyCaret on the Bank Marketing dataset.  
The goal is to predict whether a client will subscribe to a term deposit based on demographic, contact, and campaign data.

We perform the following steps:
- Data cleaning and preprocessing
- PyCaret setup and model comparison
- Final predictions and export

The output variable is `y`, where `1` indicates subscription and `0` indicates no subscription.

```python
import pandas as pd
from pycaret.classification import *
```

```python
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
```

```python
train.columns = train.columns.str.strip()  # boşluk varsa temizle
train.columns
```

```python
train.columns = train.columns.str.strip()  
s = setup(data=train, target='y', session_id=42, verbose=False)
```

```python
best_model = compare_models()
```

```python
preds = predict_model(best_model, data=test)
```

```python
submission = pd.DataFrame({'Term_Deposit_Prediction': preds['prediction_label']})
```

```python
submission.to_csv('submission.csv', index=False)
submission.head()
```

```python
save_model(best_model, 'best_automl_model') 
```
